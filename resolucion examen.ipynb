```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

def blur(image, mask, blur):
    fondo = cv2.GaussianBlur(image, (51, 51), 0)
    fin = np.where(mask[:, :, None] > 0, image, fondo)
    return fin

image = cv2.imread('segundo momento evaluativo/rosa.png')
gray = cv2.cvtColor(image, cv2.COLOR_BGR2GRAY)
mask = cv2.threshold(gray, 100, 125, cv2.THRESH_BINARY_INV)[1]
mask = cv2.bitwise_not(mask)
mask = cv2.GaussianBlur(mask, (21, 21), 0)
res = blur(image, mask, 25)

plt.figure(figsize=(15, 15))
plt.subplot(121)
plt.imshow(cv2.cvtColor(image, cv2.COLOR_BGR2RGB))
plt.title('Original')
plt.axis('off')

plt.subplot(122)
plt.imshow(cv2.cvtColor(res, cv2.COLOR_BGR2RGB))
plt.title('Blurred')
plt.axis('off')
plt.show()
```

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

image = cv2.imread('segundo momento evaluativo/elementos.jpg')
gray = cv2.cvtColor(image, cv2.COLOR_BGR2GRAY)

binario = cv2.threshold(gray, 128, 255, cv2.THRESH_BINARY_INV + cv2.THRESH_OTSU)[1]
borde = cv2.findContours(binario, cv2.RETR_EXTERNAL, cv2.CHAIN_APPROX_SIMPLE)[0]
cont = len(borde)

plt.imshow(cv2.cvtColor(image, cv2.COLOR_BGR2RGB))
plt.title(f'Elementos: {cont}')
plt.axis('off')
plt.show()
```
