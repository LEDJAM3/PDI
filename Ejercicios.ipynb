Ejercicio 1

```python
a = 5
factorial = 1

for i in range(a):
    factorial *= (i + 1)

print(factorial)
```

Ejercicio 2

```python
a = 7
aux = 0

for i in range(a):
    if a % (i + 1) == 0:
        aux += 1

if aux == 2:
    print(a, "es primo")
else:
    print(a, "no es primo")
```

Ejercicio 5

```python
a = [5, 1, 4, 1, 1, 4, 1]
contador = {}

for num in a:
    if num in contador:
        contador[num] += 1
    else:
        contador[num] = 1

numero = max(contador, key=contador.get)
frecuencia = contador[numero]

print(numero, "se repite", frecuencia, "veces")
```
