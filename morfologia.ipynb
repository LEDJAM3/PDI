USAR SKIMAGE CON EROSION, DILATACION, APERTURA Y CERRADURA

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
from skimage import morphology

imagen = cv2.imread('imagen.jpg', cv2.IMREAD_GRAYSCALE)
img_bin = cv2.threshold(imagen, 127, 255, cv2.THRESH_BINARY)[1]

kernel = morphology.disk(5)

erosion = morphology.erosion(imagen, kernel)
dilation = morphology.dilation(imagen, kernel)
apertura = morphology.opening(imagen, kernel)
cerradura = morphology.closing(imagen, kernel)

erosion_bin = morphology.binary_erosion(img_bin, kernel)
dilation_bin = morphology.binary_dilation(img_bin, kernel)
apertura_bin = morphology.binary_opening(img_bin, kernel)
cerradura_bin = morphology.binary_closing(img_bin, kernel)

plt.figure(figsize=(15, 10))
plt.subplot(331)
plt.imshow(imagen, cmap='gray')
plt.title('Imagen Original')
plt.axis('off')

plt.subplot(332)
plt.imshow(erosion, cmap='gray')
plt.title('Erosión')
plt.axis('off')

plt.subplot(333)
plt.imshow(dilation, cmap='gray')
plt.title('Dilatación')
plt.axis('off')

plt.subplot(334)
plt.imshow(apertura, cmap='gray')
plt.title('Apertura')
plt.axis('off')

plt.subplot(335)
plt.imshow(cerradura, cmap='gray')
plt.title('Cerradura')
plt.axis('off')

plt.subplot(336)
plt.imshow(erosion_bin, cmap='gray')
plt.title('Erosión Binaria')
plt.axis('off')

plt.subplot(337)
plt.imshow(dilation_bin, cmap='gray')
plt.title('Dilatación Binaria')
plt.axis('off')

plt.subplot(338)
plt.imshow(apertura_bin, cmap='gray')
plt.title('Apertura Binaria')
plt.axis('off')

plt.subplot(339)
plt.imshow(cerradura_bin, cmap='gray')
plt.title('Cerradura Binaria')
plt.axis('off')
plt.show()
```

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
from skimage.data import checkerboard

image = np.zeros((300, 300), dtype=np.uint8)

image = cv2.circle(image, (200, 100), 40, 255, -1)
image = cv2.rectangle(image, (50, 50), (100, 100), 255, -1)
image = cv2.line(image, (250, 50), (270, 200), 255, 10)

checker = checkerboard() > 0.5
checker = checker.astype(np.uint8) * 255
plt.figure(figsize=(15, 10))
plt.subplot(121)
plt.imshow(image, cmap='gray')
plt.title('Imagen Original')
plt.axis('off')

plt.subplot(122)
plt.imshow(checker, cmap='gray')
plt.title('Checkerboard')
plt.axis('off')

kernel_ellipse = cv2.getStructuringElement(cv2.MORPH_ELLIPSE, (5, 5))
kernel_cross = cv2.getStructuringElement(cv2.MORPH_CROSS, (5, 5))
kernel_rect = cv2.getStructuringElement(cv2.MORPH_RECT, (5, 5))

eroded_ellipse = cv2.erode(image, kernel_ellipse, iterations=1)
dilated_ellipse = cv2.dilate(image, kernel_ellipse, iterations=1)

eroded_cross = cv2.erode(image, kernel_cross, iterations=1)
dilated_cross = cv2.dilate(image, kernel_cross, iterations=1)

eroded_rect = cv2.erode(image, kernel_rect, iterations=1)
dilated_rect = cv2.dilate(image, kernel_rect, iterations=1)

eroded_checker_ellipse = cv2.erode(checker, kernel_ellipse, iterations=1)
dilated_checker_ellipse = cv2.dilate(checker, kernel_ellipse, iterations=1)

eroded_checker_cross = cv2.erode(checker, kernel_cross, iterations=1)
dilated_checker_cross = cv2.dilate(checker, kernel_cross, iterations=1)

eroded_checker_rect = cv2.erode(checker, kernel_rect, iterations=1)
dilated_checker_rect = cv2.dilate(checker, kernel_rect, iterations=1)

np.random.seed(42)
imagen = image.copy()
ruido = np.random.randint(0, 2, imagen.shape, dtype=np.uint8) * 255
imagen_ruidosa = cv2.bitwise_xor(imagen, ruido)

apertura_ellipse = cv2.morphologyEx(imagen_ruidosa, cv2.MORPH_OPEN, kernel_ellipse)
cerradura_ellipse = cv2.morphologyEx(imagen_ruidosa, cv2.MORPH_CLOSE, kernel_ellipse)

apertura_cross = cv2.morphologyEx(imagen_ruidosa, cv2.MORPH_OPEN, kernel_cross)
cerradura_cross = cv2.morphologyEx(imagen_ruidosa, cv2.MORPH_CLOSE, kernel_cross)

apertura_rect = cv2.morphologyEx(imagen_ruidosa, cv2.MORPH_OPEN, kernel_rect)
cerradura_rect = cv2.morphologyEx(imagen_ruidosa, cv2.MORPH_CLOSE, kernel_rect)

erosiones = [eroded_ellipse, eroded_cross, eroded_rect, eroded_checker_ellipse, eroded_checker_cross, eroded_checker_rect]
dilatasiones = [dilated_ellipse, dilated_cross, dilated_rect, dilated_checker_ellipse, dilated_checker_cross, dilated_checker_rect]
aperturas_cerraduras = [apertura_ellipse, apertura_cross, apertura_rect, cerradura_ellipse, cerradura_cross, cerradura_rect]

plt.figure(figsize=(15, 15))
for i in range(6):
    plt.subplot(6, 3, i + 1)
    plt.imshow(erosiones[i], cmap='gray')
    plt.title('Erosión' + str(i + 1))
    plt.axis('off')
for i in range(6):
    plt.subplot(6, 3, i + 7)
    plt.imshow(dilatasiones[i], cmap='gray')
    plt.title('Dilatación' + str(i + 1))
    plt.axis('off')
for i in range(6):
    plt.subplot(6, 3, i + 13)
    plt.imshow(aperturas_cerraduras[i], cmap='gray')
    if i < 3:
        plt.title('Apertura' + str(i + 1))
    else:
        plt.title('Cerradura' + str(i - 2))
    plt.axis('off')
plt.show()
```
