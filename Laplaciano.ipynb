Laplaciano

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

imagen = cv2.imread('imagen.jpg')
imagen_rgb = cv2.cvtColor(imagen, cv2.COLOR_BGR2RGB)
imagen_gray = cv2.cvtColor(imagen, cv2.COLOR_BGR2GRAY)

laplaciano = cv2.Laplacian(imagen_gray, cv2.CV_64F)

plt.figure(figsize=(10, 4))

plt.subplot(121)
plt.title("Original (RGB)")
plt.imshow(imagen_rgb)
plt.axis('off')

plt.subplot(122)
plt.title("Laplaciano")
plt.imshow(np.abs(laplaciano), cmap='gray')
plt.axis('off')

plt.tight_layout()
plt.show()
```

Laplaciano del Gaussiano

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

imagen = cv2.imread('imagen.jpg')
imagen_rgb = cv2.cvtColor(imagen, cv2.COLOR_BGR2RGB)
imagen_gray = cv2.cvtColor(imagen, cv2.COLOR_BGR2GRAY)

gauss = cv2.GaussianBlur(imagen_gray, (5, 5), sigmaX=1)

log = cv2.Laplacian(gauss, cv2.CV_64F)

plt.figure(figsize=(10, 4))

plt.subplot(1, 2, 1)
plt.title("Original (RGB)")
plt.imshow(imagen_rgb)
plt.axis('off')

plt.subplot(1, 2, 2)
plt.title("Laplaciano del Gaussiano (LoG)")
plt.imshow(np.abs(log), cmap='gray')
plt.axis('off')

plt.tight_layout()
plt.show()
```
