Imagen de 4 cuadrantes

```python
import matplotlib.pyplot as plt
import cv2
import numpy as np

plt.rcParams['figure.figsize']=[10,5]

tamanio_img=100
R = np.ones((tamanio_img, tamanio_img)) * 0
G = np.ones((tamanio_img, tamanio_img)) * 0
B = np.ones((tamanio_img, tamanio_img)) * 0

imagen=np.dstack((R,G,B))
imagen=np.array(imagen,dtype=int)

for i in range(50): 
    for j in range(50):
        imagen[i,j]=[255,0,0]

for i in range(50):
    for j in range(50):
        imagen[i,j+50]=[0,255,0]

for i in range(50):
    for j in range(50):
        imagen[i+50,j]=[0,0,255]

cv2.imwrite('img_4cuadrantes.jpg', imagen)

plt.figure()
plt.imshow(imagen)
plt.axis('off')
plt.show()
```
