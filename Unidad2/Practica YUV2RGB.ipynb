YUV2RGB y YUV2RGB

```python
import matplotlib.pyplot as plt
import cv2
import numpy as np

imagen = cv2.cvtColor(cv2.imread('Imagenjpg.jpg', cv2.IMREAD_COLOR), cv2.COLOR_BGR2RGB)

img_yuv = cv2.cvtColor(imagen, cv2.COLOR_RGB2YUV)

img_rgb = cv2.cvtColor(img_yuv, cv2.COLOR_YUV2RGB)

fig, axs = plt.subplots(1, 3, figsize = (20, 10))

axs[0].imshow(imagen)
axs[0].set_title('Original')
axs[0].axis('off')

axs[1].imshow(img_yuv)
axs[1].set_title('YUV')
axs[1].axis('off')

axs[2].imshow(img_rgb)
axs[2].set_title('RGB')
axs[2].axis('off')
```
