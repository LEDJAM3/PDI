Colores reales

```python
import cv2
import matplotlib.pyplot as plt
import numpy as np

plt.rcParams["figure.figsize"] = [15, 5]

img = cv2.imread("Imagenjpg.jpg", cv2.IMREAD_COLOR)
img_rgb = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)

r = img[:, :, 2]
g = img[:, :, 1]
b = img[:, :, 0]

r_img = np.zeros_like(img_rgb)
r_img[:, :, 0] = r

g_img = np.zeros_like(img_rgb)
g_img[:, :, 1] = g

b_img = np.zeros_like(img_rgb)
b_img[:, :, 2] = b

plt.subplot(141)
plt.title("Original")
plt.imshow(img_rgb)
plt.axis("off")

plt.subplot(142)
plt.title("Rojo")
plt.imshow(r_img)
plt.axis("off")

plt.subplot(143)
plt.title("Verde")
plt.imshow(g_img)
plt.axis("off")

plt.subplot(144)
plt.title("Azul")
plt.imshow(b_img)
plt.axis("off")

plt.show()
```

Creacion de una imagen

```python
import matplotlib.pyplot as plt
import cv2
import numpy as np

plt.rcParams['figure.figsize']=[10,5]

tamanio_img=100
R = np.ones((tamanio_img, tamanio_img)) * 255
G = np.ones((tamanio_img, tamanio_img)) * 128
B = np.ones((tamanio_img, tamanio_img)) * 0

imagen=np.dstack((R,G,B))
imagen=np.array(imagen,dtype=int)

for i in range(20):
    for j in range(20):
        imagen[i+20,j+20]=[255,0,0]

for i in range(20):
    for j in range(20):
        imagen[i+20,j+60]=[255,0,0]

for i in range(20):
    for j in range(60):
        imagen[i+60,j+20]=[255,0,0]

plt.figure()
plt.imshow(imagen)
plt.show()
```

RGB a HSV

```python
import matplotlib.pyplot as plt
import cv2
import numpy as np

fig, axs = plt.subplots(1, 4, figsize = (25,10))

img_bgr = cv2.imread('Imagenjpg.jpg')
img_rgb = cv2.cvtColor(img_bgr, cv2.COLOR_BGR2RGB)
img_hsv = cv2.cvtColor(img_bgr, cv2.COLOR_BGR2HSV)

axs[0].imshow(img_rgb)
axs[0].set_title("Imagen original")
axs[0].axis("off")

fig_img1 = axs[1].imshow(img_hsv[..., 0], cmap='gray', vmin=0, vmax=179)
fig.colorbar(fig_img1, ax=axs[1], shrink=0.3)
axs[1].set_title("H")
axs[1].axis("off")

fig_img1 = axs[2].imshow(img_hsv[..., 1], cmap='gray', vmin=0, vmax=179)
fig.colorbar(fig_img1, ax=axs[2], shrink=0.3)
axs[2].set_title("S")
axs[2].axis("off")

fig_img1 = axs[3].imshow(img_hsv[..., 2], cmap='gray', vmin=0, vmax=179)
fig.colorbar(fig_img1, ax=axs[3], shrink=0.3)
axs[3].set_title("V")
axs[3].axis("off")
```

Funciones

```python
import matplotlib.pyplot as plt
import cv2
import numpy as np
Imagen = cv2.imread('Imagenjpg.jpg')
img= cv2.cvtColor(Imagen, cv2.COLOR_BGR2RGB)
def RGB2HSV(img):
    img = img.astype(np.float32)
    img = img/255.
    new_img = np.zeros_like(img)

    maxChannel = np.argmax(img,axis=2)
    maxValue = np.amax(img,axis=2)
    minChannel = np.argmin(img,axis=2)
    minValue = np.amin(img,axis=2)
    #value 
    new_img[...,2] = maxValue

    #saturation
    for i in range(img.shape[0]):
        for j in range(img.shape[1]):
            if maxValue[i,j]==0:
                new_img[i,j,1] = 0
            elif maxValue[i,j] != 0:
                new_img[i,j,1] = (new_img[i,j,2]-minValue[i,j])/new_img[i,j,2]

    #hue 
    for i in range(img.shape[0]):
        for j in range(img.shape[1]):
            if sum(img[i,j,...])/3 == img[i,j,0]:
                new_img[i,j,0] = 0
            elif maxChannel[i,j] == 0: #R 
                new_img[i,j,0] = 60 * ((img[i,j,1]-img[i,j,2])/(new_img[i,j,2]-minValue[i,j]))
            elif maxChannel[i,j] == 1: #G
                new_img[i,j,0] = 120 + 60 * (img[i,j,2]-img[i,j,0])/(new_img[i,j,2]-minValue[i,j])
            elif maxChannel[i,j] == 2: #B
                new_img[i,j,0] = 240 + 60 * (img[i,j,0]-img[i,j,1])/(new_img[i,j,2]-minValue[i,j])

        bol_ind = new_img[...,0] < 0
        new_img[bol_ind,0]   =  new_img[bol_ind,0] + 360
    return new_img 

hsv_img =  RGB2HSV(img) 
fig, axs = plt.subplots(1,4,figsize=(25,10))
axs[0].imshow(img)
axs[0].set_title('Imagen')
fig_img = axs[1].imshow(hsv_img[...,0],cmap='gray',vmin=0,vmax=360)
fig.colorbar(fig_img, ax=axs[1],shrink=0.3)
axs[1].set_title('Canal Hue')
fig_img = axs[2].imshow(hsv_img[...,1],cmap='gray',vmin=0,vmax=1)
fig.colorbar(fig_img, ax=axs[2],shrink=0.3)
axs[2].set_title('Canal Saturación')
fig_img = axs[3].imshow(hsv_img[...,2],cmap='gray',vmin=0,vmax=1)
fig.colorbar(fig_img, ax=axs[3],shrink=0.3)
axs[3].set_title('Canal Value');
```

```python
import matplotlib.pyplot as plt
import cv2
import numpy as np

img_rgb = cv2.cvtColor(cv2.imread('Imagenjpg.jpg', cv2.IMREAD_COLOR), cv2.COLOR_BGR2RGB)

img_hsv = cv2.cvtColor(img_rgb, cv2.COLOR_RGB2HSV)

img_rgb2 = cv2.cvtColor(img_hsv, cv2.COLOR_HSV2RGB)

fig, axs = plt.subplots(1, 3, figsize = (20, 10))

axs[0].imshow(img_rgb)
axs[0].set_title('RGB 1')
axs[0].axis('off')

axs[1].imshow(img_hsv)
axs[1].set_title('HSV')
axs[1].axis('off')

axs[2].imshow(img_rgb2)
axs[2].set_title('RGB 2')
axs[2].axis('off')
```

```python
import matplotlib.pyplot as plt
import cv2
import numpy as np

imagen = cv2.imread('Imagenjpg.jpg', 0)

_, umbralBinario = cv2.threshold(imagen, 127, 255, cv2.THRESH_BINARY)

_, umbralBinarioInverso = cv2.threshold(imagen, 127, 255, cv2.THRESH_BINARY_INV)

fig, axs = plt.subplots(1, 3, figsize = (20, 10))

axs[0].imshow(imagen, cmap = 'gray')
axs[0].set_title('Original')
axs[0].axis('off')

axs[1].imshow(umbralBinario, cmap = 'gray')
axs[1].set_title('Umbral binario')
axs[1].axis('off')

axs[2].imshow(umbralBinarioInverso, cmap = 'gray')
axs[2].set_title('Umbral binario inverso')
axs[2].axis('off')
```
