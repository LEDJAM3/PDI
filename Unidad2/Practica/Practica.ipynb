```python
import cv2
import matplotlib.pyplot as plt
import numpy as np

#Funcion para mostrar imgenes
def mostrarImagenes(foto1, foto2, foto3):
    plt.rcParams["figure.figsize"] = [15, 5]
    
    plt.subplot(131)
    plt.title("Foto 1")
    plt.imshow(foto1)
    plt.axis("off")

    plt.subplot(132)
    plt.title("Foto 2")
    plt.imshow(foto2)
    plt.axis("off")

    plt.subplot(133)
    plt.title("Foto 3")
    plt.imshow(foto3)
    plt.axis("off")

    plt.show()

def grisesSubplot1(imagen):
    plt.rcParams["figure.figsize"] = [15, 5]
    
    plt.subplot(141)
    plt.title("Original")
    plt.imshow(imagen)
    plt.axis("off")

    plt.subplot(142)
    plt.title("Rojo")
    plt.imshow(imagen[:,:,0], cmap="gray")
    plt.axis("off")

    plt.subplot(143)
    plt.title("Verde")
    plt.imshow(imagen[:,:,1], cmap="gray")
    plt.axis("off")

    plt.subplot(144)
    plt.title("Azul")
    plt.imshow(imagen[:,:,2], cmap="gray")
    plt.axis("off")

    plt.show()

def grisesSubplot2(img_rgb):
    fig, axs = plt.subplots(1, 4, figsize = (20, 5))
    axs[0].imshow(img_rgb)
    axs[0].set_title("Imagen original")
    axs[0].axis("off")

    fig_img = axs[1].imshow(img_rgb[..., 0], cmap="gray")
    fig.colorbar(fig_img, ax=axs[1], shrink = 0.5)
    axs[1].set_title("Rojo")
    axs[1].axis("off")

    fig_img = axs[2].imshow(img_rgb[..., 1], cmap="gray")
    fig.colorbar(fig_img, ax=axs[2], shrink = 0.5)
    axs[2].set_title("Verde")
    axs[2].axis("off")

    fig_img = axs[3].imshow(img_rgb[..., 2], cmap="gray")
    fig.colorbar(fig_img, ax=axs[3], shrink = 0.5)
    axs[3].set_title("Azul")
    axs[3].axis("off")

def coloresRealesSubplot1(img, img_rgb):
    r = img[:, :, 2]
    g = img[:, :, 1]
    b = img[:, :, 0]

    r_img = np.zeros_like(img_rgb)
    r_img[:, :, 0] = r

    g_img = np.zeros_like(img_rgb)
    g_img[:, :, 1] = g

    b_img = np.zeros_like(img_rgb)
    b_img[:, :, 2] = b

    plt.subplot(141)
    plt.title("Original")
    plt.imshow(img_rgb)
    plt.axis("off")

    plt.subplot(142)
    plt.title("Rojo")
    plt.imshow(r_img)
    plt.axis("off")

    plt.subplot(143)
    plt.title("Verde")
    plt.imshow(g_img)
    plt.axis("off")

    plt.subplot(144)
    plt.title("Azul")
    plt.imshow(b_img)
    plt.axis("off")

    plt.show()

def coloresRealesSubplot2(img, img_rgb):
    r = img[:, :, 2]
    g = img[:, :, 1]
    b = img[:, :, 0]

    r_img = np.zeros_like(img_rgb)
    r_img[:, :, 0] = r

    g_img = np.zeros_like(img_rgb)
    g_img[:, :, 1] = g

    b_img = np.zeros_like(img_rgb)
    b_img[:, :, 2] = b

    fig, axs = plt.subplots(1, 4, figsize = (20, 5))
    axs[0].imshow(img_rgb)
    axs[0].set_title("Imagen original")
    axs[0].axis("off")

    fig_img = axs[1].imshow(r_img)
    fig.colorbar(fig_img, ax=axs[1], shrink = 0.5)
    axs[1].set_title("Rojo")
    axs[1].axis("off")

    fig_img = axs[2].imshow(g_img)
    fig.colorbar(fig_img, ax=axs[2], shrink = 0.5)
    axs[2].set_title("Verde")
    axs[2].axis("off")

    fig_img = axs[3].imshow(b_img)
    fig.colorbar(fig_img, ax=axs[3], shrink = 0.5)
    axs[3].set_title("Azul")
    axs[3].axis("off")

def coloresFalsosSubplot1(img, img_rgb):
    plt.rcParams["figure.figsize"] = [15, 5]

    b, g, r = cv2.split(img)

    plt.subplot(141)
    plt.title("Original")
    plt.imshow(img_rgb)
    plt.axis("off")

    plt.subplot(142)
    plt.title("Rojo")
    plt.imshow(r, cmap='Reds')
    plt.axis("off")

    plt.subplot(143)
    plt.title("Verde")
    plt.imshow(g, cmap='Greens')
    plt.axis("off")

    plt.subplot(144)
    plt.title("Azul")
    plt.imshow(b, cmap='Blues')
    plt.axis("off")

    plt.show()

def coloresFalsosSubplot2(img, img_rgb):
    b, g, r = cv2.split(img)

    fig, axs = plt.subplots(1, 4, figsize = (20, 5))
    axs[0].imshow(img_rgb)
    axs[0].set_title("Imagen original")
    axs[0].axis("off")

    fig_img = axs[1].imshow(r, cmap='Reds')
    fig.colorbar(fig_img, ax=axs[1], shrink = 0.5)
    axs[1].set_title("Rojo")
    axs[1].axis("off")

    fig_img = axs[2].imshow(g, cmap='Greens')
    fig.colorbar(fig_img, ax=axs[2], shrink = 0.5)
    axs[2].set_title("Verde")
    axs[2].axis("off")

    fig_img = axs[3].imshow(b, cmap='Blues')
    fig.colorbar(fig_img, ax=axs[3], shrink = 0.5)
    axs[3].set_title("Azul")
    axs[3].axis("off")
```

```python
import cv2
import matplotlib.pyplot as plt
import numpy as np

#Cargar las imagenes
foto1_bgr = cv2.imread("foto1.jpeg", cv2.IMREAD_COLOR)
foto2_bgr = cv2.imread("foto2.jpeg", cv2.IMREAD_COLOR)
foto3_bgr = cv2.imread("foto3.jpeg", cv2.IMREAD_COLOR)

#imagenes convertidas a rgb
foto1_rgb = cv2.cvtColor(foto1_bgr, cv2.COLOR_BGR2RGB)
foto2_rgb = cv2.cvtColor(foto2_bgr, cv2.COLOR_BGR2RGB)
foto3_rgb = cv2.cvtColor(foto3_bgr, cv2.COLOR_BGR2RGB)

#guardar imagenes
cv2.imwrite("rgb_guardada1.jpeg", foto1_bgr)
cv2.imwrite("rgb_guardada2.jpeg", foto2_rgb)
cv2.imwrite("rgb_guardada3.jpeg", foto3_rgb)

cv2.imwrite("bgr_guardada1.jpeg", foto1_bgr)
cv2.imwrite("bgr_guardada2.jpeg", foto2_bgr)
cv2.imwrite("bgr_guardada3.jpeg", foto3_bgr)

#mostrar imagenes
mostrarImagenes(foto1_rgb, foto2_rgb, foto3_rgb)
```

```python
#Imagenes en grises con cv2
gris1 = cv2.cvtColor(foto1_bgr, cv2.COLOR_BGR2GRAY)
gris2 = cv2.cvtColor(foto2_bgr, cv2.COLOR_BGR2GRAY)
gris3 = cv2.cvtColor(foto3_bgr, cv2.COLOR_BGR2GRAY)

plt.rcParams["figure.figsize"] = [15, 5]

plt.subplot(131)
plt.title("Foto1")
plt.imshow(gris1, cmap="gray")
plt.axis("off")

plt.subplot(132)
plt.title("Foto2")
plt.imshow(gris2, cmap="gray")
plt.axis("off")

plt.subplot(133)
plt.title("Foto3")
plt.imshow(gris3, cmap="gray")
plt.axis("off")

plt.show()
```

Escala de grises Subplot 1

```python
grisesSubplot1(foto1_rgb)
grisesSubplot1(foto2_rgb)
grisesSubplot1(foto3_rgb)
```

Escala de grises subplot 2

```python
grisesSubplot2(foto1_rgb)
grisesSubplot2(foto2_rgb)
grisesSubplot2(foto3_rgb)
```

Colores reales subplot 1

```python
coloresRealesSubplot1(foto1_bgr, foto1_rgb)
coloresRealesSubplot1(foto2_bgr, foto2_rgb)
coloresRealesSubplot1(foto3_bgr, foto3_rgb)
```

Colores reales subplot 2

```python
coloresRealesSubplot2(foto1_bgr, foto1_rgb)
coloresRealesSubplot2(foto2_bgr, foto2_rgb)
coloresRealesSubplot2(foto3_bgr, foto3_rgb)
```

Colores falsos subplot 1

```python
coloresFalsosSubplot1(foto1_bgr, foto1_rgb)
coloresFalsosSubplot1(foto2_bgr, foto2_rgb)
coloresFalsosSubplot1(foto3_bgr, foto3_rgb)
```

```python
coloresFalsosSubplot2(foto1_bgr, foto1_rgb)
coloresFalsosSubplot2(foto2_bgr, foto2_rgb)
coloresFalsosSubplot2(foto3_bgr, foto3_rgb)
```
