```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

def convMediaHor(imagen, tamanioK):
    kernel = np.ones((1, tamanioK), np.float32) / tamanioK
    imgFin = cv2.filter2D(imagen, -1, kernel)
    return imgFin

imagen = cv2.imread("imagen distorsionada.jpeg")

gris = cv2.cvtColor(imagen, cv2.COLOR_BGR2GRAY)

img = convMediaHor(imagen, 25)

plt.figure(figsize=(10, 5))
plt.subplot(121)
plt.imshow(cv2.cvtColor(imagen, cv2.COLOR_BGR2RGB))

plt.subplot(122)
plt.imshow(img, cmap='gray')
plt.show()
```

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

def convMediaHor(imagen, tamanioK):
    kernel = np.ones((tamanioK, 1), np.float32) / tamanioK
    imgFin = cv2.filter2D(imagen, -1, kernel)
    return imgFin

imagen = cv2.imread("img vertical.jpg")

gris = cv2.cvtColor(imagen, cv2.COLOR_BGR2RGB)

img = convMediaHor(gris, 50)

plt.figure(figsize=(10, 5))
plt.subplot(121)
plt.imshow(cv2.cvtColor(imagen, cv2.COLOR_BGR2RGB))

plt.subplot(122)
plt.imshow(img)
plt.show()
```
